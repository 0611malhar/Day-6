```python
# Write a program using functions to find greatest of three numbers.
def got(a,b,c):
    if a>=b and a>=c:
        print(a,"is greatest")
    elif b>=a and b>=c:
        print(b,"is greatest")
    else:
        print(c,"is greatest")
a=int(input())
b=int(input())
c=int(input())
got(a,b,c)
```

```python
# Write a python program using function to convert Celsius to Fahrenheit.

def got(c):
    print(32+((9/5)*c))
c=int(input())
got(c)
```

```python
# Write a recursive function to calculate the sum of first n natural numbers.

def got(c):
    if c==1:
        return 1
    return c+got(c-1)
c=int(input())
print(got(c))
```
