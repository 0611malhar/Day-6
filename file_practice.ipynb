```python
lines = 0
words = 0
spaces = 0
characters = 0

f = open('abd.txt')

in_word = False

while True:
    ch = f.read(1)

    if ch == '':# EOF
        break

    if ch == '\n':
        lines += 1

    if ch == ' ':
        spaces += 1

    if ch != ' ' and ch != '\n':
        characters += 1

    if ch != ' ' and ch != '\n' and in_word == False:
        words += 1
        in_word = True

    if ch == ' ' or ch == '\n':
        in_word = False

f.close()

print("Lines =", lines)
print("Words =", words)
print("Spaces =", spaces)
print("Characters =", characters)
```
