```python
import numpy as np
```

```python
cli = np.genfromtxt('climate.txt',delimiter= ',', skip_header= 1)# txt to np array
```

```python
cli
```

```python
wt = np.array([0.3,0.2,0.5])
```

```python
# @ -> matrix * operator, both need tobe in the form of np.array
y = cli @ wt
y
```

```python
# append it back to climate.txt file
c_r = np.concatenate((cli,y.reshape(100,1)) , axis = 1)
```

```python
c_r
```

```python
np.savetxt('cr.txt',
          c_r,
          fmt='%.2f',
          header='temperature, rainfall, humidity, yeild_apples',
          comments='')
```
