```python
'''
Read the file line by line
Parse the first line to get list of the column names or headers
Split each remaining line and convert each value into a float
Create a dictionary for each loan using the headers as keys
Create a list of dictionaries to keep track of all the loans
'''

f3 = open('./data/loan3.txt' , mode = 'r')
f3ls = f3.readlines()
print(f3ls)
```

```python
def parse_h(h):
    return h.strip().split(',')

print(parse_h(f3ls[0]))
```

```python
def x(d):
    v = []
    for i in d.strip().split(','):
        if i=='':
            v.append(0.0)
        else:
            v.append(float(i))
    return v
```

```python
print(x(f3ls[2]))
```

```python
def create_item_dict(values, headers):
    result = {}
    for value, header in zip(values, headers):
        result[header] = value
    return result
```

```python
def read_csv(path) :
    result = []
    with open(path, 'r') as f:
        lines = f.readlines ()
        headers = parse_h(lines [ 0])
        for data_line in lines[1:]:
            values = x(data_line)
            item_dict = create_item_dict(values, headers)
            result.append(item_dict)
    return result
```

```python
read_csv('./data/loan2.txt')
```

```python
def compute_emis(loans) :
    for loan in loans:
        loan['emi' ] = loan_emi (
            loan['amount' ],
            loan['duration' ],
            loan['rate' ]/12, # the CSV contains yearly rates
            loan['down_payment'])
```

```python
import math

def loan_emi(amount, duration, rate, down_payment=0):
    """Calculates the equal montly installment (EMI) for a loan.
I

    amount - Total amount to be spent (loan + down payment)
    duration - Duration of the loan (in months)
    rate - Rate of interest (monthly)
    down_payment (optional) - Optional intial payment (deducted from ar
    """
    loan_amount = amount - down_payment
    try:
        emi = loan_amount * rate * ((1+rate) ** duration) / (((1+rate) ** duration))
    except ZeroDivisionError:
        emi = loan_amount / duration
    emi = math.ceil(emi)
    return emi
```

```python
def write_csv(items, path):
    # Open the file in write mode
    with open(path, 'w') as f:
        # Return if there's nothing to write
        if len(items) == 0:
            return

        # Write the headers in the first line
        headers = list(items[0].keys())
        f.write(','.join(headers) + '\n')

        # Write one item per line
        for item in items:
            values = []
            for header in headers:
                values.append(str(item.get(header, "")))
                f.write(','.join(values) + "\n")
```
